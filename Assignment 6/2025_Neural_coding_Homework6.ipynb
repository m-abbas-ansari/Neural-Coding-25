## Exercise 1

Let random variable $X$ be a discrete random variable with values
{−2, −1, 0, 1, 2}, obtained with the same probability $p = \tfrac15$. Let
the second random variable $Y$ be $Y = (X+1)^2$.

(a) Compute analytically $cov(X, Y)$ (2 points)


We are given:

$$
X \in \{-2, -1, 0, 1, 2\}, \quad p(x) = \frac{1}{5}, \quad Y = (X + 1)^2
$$

We use the identity:

$$
\mathrm{Cov}(X, Y) = \mathbb{E}[XY] - \mathbb{E}[X] \mathbb{E}[Y]
$$

First, compute the expectation $\mathbb{E}[X]$:

$$
\mathbb{E}[X] = \sum_{x} p(x) \cdot x = \frac{1}{5}(-2 - 1 + 0 + 1 + 2) = 0
$$

Next, compute $\mathbb{E}[Y]$:

$$
\mathbb{E}[Y] = \sum_{x} p(x) \cdot (x + 1)^2 = \frac{1}{5}\left[(−2+1)^2 + (−1+1)^2 + (0+1)^2 + (1+1)^2 + (2+1)^2\right]
= \frac{1}{5}(1 + 0 + 1 + 4 + 9) = 3
$$

Now compute $\mathbb{E}[XY]$:

$$
\mathbb{E}[XY] = \sum_{x} p(x) \cdot x \cdot (x + 1)^2 = \frac{1}{5} \left[ (-2)(-1)^2 + (-1)(0)^2 + 0(1)^2 + 1(2)^2 + 2(3)^2 \right]
$$

$$
= \frac{1}{5} \left[ -2 + 0 + 0 + 4 + 18 \right] = \frac{20}{5} = 4
$$

Thus:

$$
\mathrm{Cov}(X, Y) = 4 - 0 \cdot 3 = \boxed{4}
$$



(b) Compute the mutual information between X and Y in bits (2 points).



We use:

$$
I(X; Y) = \sum_{x,y} p(x, y) \log_2 \left( \frac{p(x, y)}{p(x)p(y)} \right)
$$

Since $X$ is uniform, $p(x) = \frac{1}{5}$. Each $x$ maps to exactly one $y = (x + 1)^2$, so:

$$
p(x, y) = p(x), \quad p(y|x) = 1
$$

We compute $p(y)$ as:

$$
p(y = 0) = \frac{1}{5}, \quad p(y = 1) = \frac{2}{5}, \quad p(y = 4) = \frac{1}{5}, \quad p(y = 9) = \frac{1}{5}
$$

So mutual information becomes:

$$
I(X; Y) = \sum_{x} p(x) \log_2 \left( \frac{1}{p(y(x))} \right)
$$

Where $y(x) = (x+1)^2$. Then:

$$
I(X; Y) = \frac{1}{5} \left[ \log_2\left(\frac{1}{2/5}\right) + \log_2\left(\frac{1}{1/5}\right) + \log_2\left(\frac{1}{2/5}\right) + \log_2\left(\frac{1}{1/5}\right) + \log_2\left(\frac{1}{1/5}\right) \right]
$$

$$
= \frac{1}{5} \left[ 2\log_2\left(\frac{5}{2}\right) + 3\log_2(5) \right]
= \frac{1}{5} \left[ 2(\log_2(5) - \log_2(2)) + 3\log_2(5) \right]
$$

$$
= \frac{1}{5} \left[ 5\log_2(5) - 2 \right] = \log_2(5) - \frac{2}{5}
$$

Using $\log_2(5) \approx 2.3219$, we get:

$$
I(X; Y) \approx 2.3219 - 0.4 = \boxed{1.9219 \text{ bits}}
$$



(c) Simulate 100, 1000, 10000 data realization of these processes and compute the covariances and mutual information based on the data. Compute the mutual information both with a library and yourself using the formula. (3 points)

```python
import numpy as np
from npeet.entropy_estimators import mi # Non-parametric Entropy Estimation Toolbox
from collections import Counter

np.random.seed(42)  # For reproducibility

def simulate(n_samples):
    # Define the random variable X
    X_vals = np.array([-2, -1, 0, 1, 2])
    X = np.random.choice(X_vals, size=n_samples, replace=True)
    Y = (X + 1) ** 2

    # Compute empirical covariance
    cov_empirical = np.cov(X, Y, ddof=0)[0, 1]

    # Mutual information using npeet (library)
    mi_lib = mi(X, Y)

    # Mutual information manually
    joint_counts = Counter(zip(X, Y))
    px_counts = Counter(X)
    py_counts = Counter(Y)
    mi_manual = 0
    for (x, y), joint_count in joint_counts.items():
        p_xy = joint_count / n_samples
        p_x = px_counts[x] / n_samples
        p_y = py_counts[y] / n_samples
        mi_manual += p_xy * np.log2(p_xy / (p_x * p_y))

    return cov_empirical, mi_lib, mi_manual

for n in [100, 1000, 10000]:
    cov, mi_lib, mi_manual = simulate(n)
    print(f"Samples: {n}")
    print(f"  Covariance: {cov:.4f}")
    print(f"  MI (npeet): {mi_lib:.4f}")
    print(f"  MI (manual): {mi_manual:.4f}")
    print("-" * 30)
```

## Exercise 2
Suppose that we have a neuron which, in a given time period, will fire
with probability 0.1, yielding a Bernoulli distribution for the neuron’s firing (denoted by the random variable R = 0 or 1) with p(R = 1) = 0.2.

(a) Compute the entropy H(R) of this distribution (calculated in bits, i.e., using the base 2 logarithm)? (1 point)

We are given a Bernoulli random variable $R \in \{0, 1\}$ with:

$$
p(R = 1) = 0.2, \quad p(R = 0) = 0.8
$$

The entropy is given by:

$$
H(R) = -\sum_{r \in \{0,1\}} p(r) \log_2 p(r) = -[p(0) \log_2 p(0) + p(1) \log_2 p(1)]
$$

$$
= -[0.8 \log_2(0.8) + 0.2 \log_2(0.2)]
$$

Using:

$$
\log_2(0.8) \approx -0.3219, \quad \log_2(0.2) \approx -2.3219
$$

We get:

$$
H(R) = -[0.8 \cdot (-0.3219) + 0.2 \cdot (-2.3219)] = 0.8 \cdot 0.3219 + 0.2 \cdot 2.3219
$$

$$
= 0.2575 + 0.4644 = \boxed{0.7219 \ \text{bits}}
$$

(b) Now lets add a stimulus to the picture. Suppose that we think this
neuron's activity is related to a light flashing in the eye. Let us say that
the light is flashing in a given time period with probability 0.2. Call this
stimulus random variable S. If there is a flash, the neuron will fire with
probability 1/2. If there is no flash, the neuron will fire with probability
1/8. Call the random variable describing whether the neuron fires or not
R. Compute the mutual information I(S : R)? (3 points)\
*Hint: First, confirm that H(R) is the same as above by computing p(R).*

Let $S \in \{0, 1\}$ be the stimulus (flash) variable, with:

$$
p(S = 1) = 0.2, \quad p(S = 0) = 0.8
$$

Firing probabilities given stimulus:

$$
p(R = 1 \mid S = 1) = \frac{1}{2}, \quad p(R = 1 \mid S = 0) = \frac{1}{8}
$$

We first compute the marginal $p(R = 1)$ using the law of total probability:

$$
p(R = 1) = p(S = 1) p(R = 1 \mid S = 1) + p(S = 0) p(R = 1 \mid S = 0)
$$

$$
= 0.2 \cdot \frac{1}{2} + 0.8 \cdot \frac{1}{8} = 0.1 + 0.1 = 0.2
$$

Thus, $p(R = 1) = 0.2$, $p(R = 0) = 0.8$, and entropy $H(R) = 0.7219 \ \text{bits}$ is confirmed.

Now we compute the **conditional entropy** $H(R \mid S)$:

$$
H(R \mid S) = p(S = 0) H(R \mid S = 0) + p(S = 1) H(R \mid S = 1)
$$

Compute each:

* For $S = 0$, we have:

$$
p(R = 1 \mid S = 0) = \frac{1}{8}, \quad p(R = 0 \mid S = 0) = \frac{7}{8}
$$

$$
H(R \mid S = 0) = -\left( \frac{1}{8} \log_2 \frac{1}{8} + \frac{7}{8} \log_2 \frac{7}{8} \right)
$$

$$
= -\left( \frac{1}{8} \cdot (-3) + \frac{7}{8} \cdot \log_2(0.875) \right)
$$

$$
\log_2(0.875) \approx -0.1926 \Rightarrow H(R \mid S = 0) \approx \frac{3}{8} + \frac{7}{8} \cdot 0.1926 = 0.375 + 0.1685 = 0.5435
$$

* For $S = 1$, we have:

$$
p(R = 1 \mid S = 1) = \frac{1}{2}, \quad p(R = 0 \mid S = 1) = \frac{1}{2}
$$

$$
H(R \mid S = 1) = -\left( \frac{1}{2} \log_2 \frac{1}{2} + \frac{1}{2} \log_2 \frac{1}{2} \right) = 1
$$

Now plug in:

$$
H(R \mid S) = 0.8 \cdot 0.5435 + 0.2 \cdot 1 = 0.4348 + 0.2 = 0.6348
$$

Now compute mutual information:

$$
I(S : R) = H(R) - H(R \mid S) = 0.7219 - 0.6348 = \boxed{0.0871 \ \text{bits}}
$$
